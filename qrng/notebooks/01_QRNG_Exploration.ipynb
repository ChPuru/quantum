## Dice

`random_int` draws three bits for a six-sided die and throws away 6 and 7, so every face stays equally likely.

```python
rolls = Counter(random_int(1, 6, seed=s) for s in range(6000))
plt.bar(sorted(rolls), [rolls[k] for k in sorted(rolls)])
plt.axhline(1000, color="k", linestyle="--")
plt.xlabel("face")
plt.ylabel("count");
```
